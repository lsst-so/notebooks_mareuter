```python
import holoviews as hv
import hvplot.pandas
import numpy as np
import pandas as pd
```

```python
dayObs = 20260112
filter_string = 'ISR start time (shutter)'
# imageTypes = ['acq', 'science', 'cwfs', 'engtest', 'bias', 'dark', 'flat']
perfTable = pd.read_json(f'/project/rubintv/raPerformance/sidecar_metadata/dayObs_{dayObs}.json').T
perfTable = perfTable.sort_index()
# print(perfTable.head)
imageTable = pd.read_json(f'/project/rubintv/LSSTCam/sidecar_metadata/dayObs_{dayObs}.json').T
imageTable = imageTable.sort_index()
# print(perfTable.head)
mergedTable = pd.merge(imageTable, perfTable, left_index=True, right_index=True, how='outer')
# mergedTable = mergedTable[mergedTable['Image type_x'].isin(imageTypes)]
mergedTable['DateBeg'] = pd.to_datetime(mergedTable['Date begin'],format="%Y-%m-%dT%H:%M:%S.%f")
mergedTable[filter_string] = mergedTable[filter_string].astype(float)
mergedTable = mergedTable[mergedTable[filter_string] < 9999998]

# print(mergedTable.head)
mt = mergedTable.hvplot.scatter("DateBeg", filter_string, by=["Image type_x"], hover_cols=["Image type_x", "index"], legend=True, title=f'{dayObs}')
mt
```

```python
# print(f"RA_Performance_{dayObs}")
hvplot.save(mt, f"~/RA/RA_Performance_{dayObs}.html")
# hvplot.save(mt, f"~/RA/RA_Performance_{dayObs}.png")
```
