```python
import os

from astropy.io import fits
from astropy.utils.data import download_file
from astropy.visualization import astropy_mpl_style
import matplotlib.pyplot as plt
import numpy as np

from lsst.sitcom.mareuter.site_lfa import get_lfa
from lsst.sitcom.mareuter.helpers import get_dir_from_fs_file

plt.style.use(astropy_mpl_style)
```

```python
intensity_data = None
wavelength_data = None
primary_header = None
extension_header = None
use_fsspec = False
fsspec_kwargs = None

fs_file = "FiberSpectrograph:Blue_fiberSpecBlue_2026-07-14T15:05:40.032.fits"
file_dir = get_dir_from_fs_file(fs_file)
lfa = get_lfa()

if lfa.profile == "lfa":
    os.environ["AWS_PROFILE"] = "lfa"
    use_fsspec = True
    fsspec_kwargs = {
        "endpoint_url": lfa.endpoint_url
    }
    url = f"s3://{lfa.bucket}/{file_dir}/{fs_file}"
else:
    url = f"{lfa.endpoint_url}/{lfa.bucket}/{file_dir}/{fs_file}"
    if "AWS_PROFILE" in os.environ:
        del os.environ["AWS_PROFILE"]

print(url)
```

```python
with fits.open(url, use_fsspec=use_fsspec, fsspec_kwargs=fsspec_kwargs) as hdul:
    # print(hdul.info())
    intensity_data = hdul[0].data
    primary_header = hdul[0].header
    wavelength_data = hdul[1].data
    extension_header = hdul[1].header
```

```python
x = wavelength_data[0][0].flatten()
```
