```python
import asyncio
import logging
import os
import sys

from lsst.ts import salobj
```

```python
stream_handler = logging.StreamHandler(sys.stdout)

logger = logging.getLogger()
logger.addHandler(stream_handler)
logger.level = logging.WARNING
```

```python
def print_event(name):
    def callback(data):
        print(name)
        print(data)
        print(os.linesep)
    return callback
```

```python
async with salobj.Domain() as domain:
    await asyncio.sleep(10)
    mtcam = salobj.Remote(domain, 'MTCamera')
    await mtcam.start_task
    mtcam.tel_utiltrunk_UT.callback = print_event("utiltrunk_UT")

    while True:
        await asyncio.sleep(5.0)
```
